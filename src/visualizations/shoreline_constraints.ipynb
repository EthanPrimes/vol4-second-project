```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
north = pd.read_csv('../data/north_shoreline.csv')
south = pd.read_csv('../data/south_shoreline.csv')
```

```python
x1, y1 = north['x'].to_numpy(), north['y'].to_numpy()

poly = np.polynomial.Chebyshev.fit(y1, x1, deg=7)
y_interp = np.linspace(44.95, 45.4, 100)
x_interp = poly(y_interp)

plt.scatter(x1, y1)
plt.plot(x_interp, y_interp)
plt.show()
```

```python
x1, y1 = south['x'].to_numpy(), south['y'].to_numpy()
plt.scatter(x1, y1)
```

```python
x1, y1 = south['x'].to_numpy(), south['y'].to_numpy()

poly = np.polynomial.Chebyshev.fit(x1, y1, deg=9)
x_interp = np.linspace(-87.7, -87.15, 100)
y_interp = poly(x_interp)

plt.scatter(x1, y1)
plt.plot(x_interp, y_interp)
plt.show()
```
